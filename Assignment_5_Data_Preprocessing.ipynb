```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/GeethaGunasekaran1/Dataset_rep/refs/heads/main/diabetes.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("=" * 60)

print("\nFIRST 5 ROWS:")
display(df.head())

print("\nLAST 5 ROWS:")
display(df.tail())

print("\nDATASET SHAPE:")
print(df.shape)

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nDATASET INFORMATION:")
df.info()

print("\nDATA TYPES:")
print(df.dtypes)

numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns

print("\nNUMERICAL COLUMNS:")
print(list(numerical_columns))

categorical_columns = df.select_dtypes(
    include=['object', 'category']
).columns

print("\nCATEGORICAL COLUMNS:")
print(list(categorical_columns))
```

#Task 2 – Data Cleaning
● Rename Columns: Update column names such as 'ID' to 'Visit_ID' and 'No_Pation' to
'Patient_ID' to make them more descriptive and clear.
● Check Unique Values in Categorical Columns: Identify the unique values in categorical
columns:
○ Gender: F (Female) & M (Male),
○ CLASS: N (No diabetes), P (Pre-diabetes), Y (Yes, Diabetes).
Review and correct any unexpected or incorrect values if necessary.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_clean = df.copy()

df_clean.rename(columns={
    'ID': 'Visit_ID',
    'No_Pation': 'Patient_ID'
}, inplace=True)

if 'Unnamed: 0' in df_clean.columns:
    df_clean.drop(columns=['Unnamed: 0'], inplace=True)

print("Updated Columns:")
print(df_clean.columns.tolist())

categorical_columns = ['Gender', 'CLASS']

for col in categorical_columns:
    print(f"\nUnique values in {col}:")
    print(df_clean[col].unique())

    print(f"\nValue counts for {col}:")
    print(df_clean[col].value_counts(dropna=False))

df_clean['Gender'] = (
    df_clean['Gender']
    .astype('string')
    .str.strip()
    .str.upper()
)

df_clean['CLASS'] = (
    df_clean['CLASS']
    .astype('string')
    .str.strip()
    .str.upper()
)

valid_gender = ['F', 'M']
valid_class = ['N', 'P', 'Y']

invalid_gender = df_clean[
    ~df_clean['Gender'].isin(valid_gender) &
    df_clean['Gender'].notna()
]

invalid_class = df_clean[
    ~df_clean['CLASS'].isin(valid_class) &
    df_clean['CLASS'].notna()
]

print("\nInvalid Gender values:")
print(invalid_gender['Gender'].unique())

print("\nInvalid CLASS values:")
print(invalid_class['CLASS'].unique())

clinical_numeric_columns = [
    'AGE',
    'Urea',
    'Cr',
    'HbA1c',
    'Chol',
    'TG',
    'HDL',
    'LDL',
    'VLDL',
    'BMI'
]

print("\nClinical numerical columns:")
print(clinical_numeric_columns)

statistical_summary = pd.DataFrame({
    'Mean': df_clean[clinical_numeric_columns].mean(),
    'Median': df_clean[clinical_numeric_columns].median(),
    'Minimum': df_clean[clinical_numeric_columns].min(),
    'Maximum': df_clean[clinical_numeric_columns].max(),
    'Standard Deviation': df_clean[clinical_numeric_columns].std()
})

print("\nStatistical Summary:")
display(statistical_summary)

plt.figure(figsize=(16, 8))

df_clean[clinical_numeric_columns].boxplot()

plt.title('Box Plot of Clinical Numerical Variables')
plt.xlabel('Clinical Variables')
plt.ylabel('Values')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()


missing_summary = pd.DataFrame({
    'Missing Count': df_clean.isnull().sum(),
    'Missing Percentage':
        df_clean.isnull().sum() / len(df_clean) * 100
})

print("\nMissing Value Summary:")
display(missing_summary)

print("\nColumns containing missing values:")
display(
    missing_summary[
        missing_summary['Missing Count'] > 0
    ]
)

for col in clinical_numeric_columns:

    if df_clean[col].isnull().sum() > 0:

        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

for col in ['Gender', 'CLASS']:

    if df_clean[col].isnull().sum() > 0:

        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )


total_missing = df_clean.isnull().sum().sum()

print("\nTotal missing values after imputation:")
print(total_missing)

if total_missing == 0:
    print("SUCCESS: No missing values remain.")
else:
    print("WARNING: Missing values still exist.")


duplicate_count = df_clean.duplicated().sum()

print("\nNumber of duplicate rows:")
print(duplicate_count)


rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates()

rows_after = len(df_clean)

duplicates_removed = rows_before - rows_after

print("\nRows before duplicate removal:", rows_before)
print("Rows after duplicate removal:", rows_after)
print("Duplicates removed:", duplicates_removed)

print("\nFINAL DATASET INFORMATION")
print("=" * 50)

print("Shape:", df_clean.shape)

print("Missing values:",
      df_clean.isnull().sum().sum())

print("Duplicate rows:",
      df_clean.duplicated().sum())

print("\nFinal columns:")
print(df_clean.columns.tolist())

print("\nFinal categorical values:")

print("Gender:",
      df_clean['Gender'].unique())

print("CLASS:",
      df_clean['CLASS'].unique())
```

#Task 3 – Outlier Handling
● Retain Outliers: Retain outliers in the AGE, HbA1c, and BMI columns, explaining why
they should be preserved.
● Filter Outliers: In the Creatinine Ratio (Cr) column, exclude values that exceed the
99.5th percentile threshold. Similarly, in the Urea column, exclude values that exceed
the 99.9th percentile threshold. Use these specific percentile thresholds for filtering,
and do not apply IQR or other methods.
● Remove Extreme Outliers: For lipid-related columns (LDL, VLDL, HDL, TG, Chol), remove
extreme outliers using an appropriate method such as IQR or Z-score.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

retain_outlier_columns = [
    'AGE',
    'HbA1c',
    'BMI'
]

print("Outliers retained in:")
print(retain_outlier_columns)

cr_threshold = df_clean['Cr'].quantile(0.995)

cr_outliers = (
    df_clean['Cr'] > cr_threshold
).sum()

print("\nCr 99.5th percentile:", cr_threshold)
print("Cr outliers:", cr_outliers)

rows_before = len(df_clean)

df_clean = df_clean[
    df_clean['Cr'] <= cr_threshold
].copy()

print("Cr rows removed:",
      rows_before - len(df_clean))


urea_threshold = df_clean['Urea'].quantile(0.999)

urea_outliers = (
    df_clean['Urea'] > urea_threshold
).sum()

print("\nUrea 99.9th percentile:", urea_threshold)
print("Urea outliers:", urea_outliers)

rows_before = len(df_clean)

df_clean = df_clean[
    df_clean['Urea'] <= urea_threshold
].copy()

print("Urea rows removed:",
      rows_before - len(df_clean))


lipid_columns = [
    'LDL',
    'VLDL',
    'HDL',
    'TG',
    'Chol'
]

outlier_summary = []

for col in lipid_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    )

    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outliers': outliers.sum()
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

print("\nLipid Outlier Summary:")
display(outlier_summary_df)

rows_before_lipid = len(df_clean)

for col in lipid_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower_bound) &
        (df_clean[col] <= upper_bound)
    ].copy()

rows_after_lipid = len(df_clean)

print("\nRows before lipid filtering:",
      rows_before_lipid)

print("Rows after lipid filtering:",
      rows_after_lipid)

print("Lipid outlier rows removed:",
      rows_before_lipid - rows_after_lipid)

print("\nFINAL DATASET AFTER TASK 3")
print("=" * 50)

print("Shape:", df_clean.shape)

print("\nAGE range:")
print(df_clean['AGE'].min(), "to", df_clean['AGE'].max())

print("\nHbA1c range:")
print(df_clean['HbA1c'].min(), "to", df_clean['HbA1c'].max())

print("\nBMI range:")
print(df_clean['BMI'].min(), "to", df_clean['BMI'].max())

print("\nMaximum Cr:")
print(df_clean['Cr'].max())

print("\nMaximum Urea:")
print(df_clean['Urea'].max())
```

#Task 4 – Data Transformation
● Feature Engineering: Apply Label Encoding or One-Hot Encoding to the categorical
column Gender to convert it into numeric format for model-building purposes.
● Feature Scaling: Standardize or normalize the numerical columns that have varying
scales (e.g., Patients, AGE, BMI, Cr) to ensure consistent contribution in subsequent
analyses. Justify the scaling method used (standardization vs. normalization).

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

print("Current columns:")
print(df_clean.columns.tolist())

if 'Gender' not in df_clean.columns:
    print("\nERROR: Gender column is already removed/encoded.")
    print("Restart from Task 2/Task 3 and run Task 4 only once.")

else:


    print("\nGender values:")
    print(df_clean['Gender'].value_counts(dropna=False))


    encoder = OneHotEncoder(
        drop='if_binary',
        sparse_output=False,
        handle_unknown='ignore'
    )

    gender_encoded = encoder.fit_transform(
        df_clean[['Gender']]
    )

    encoded_columns = encoder.get_feature_names_out(
        ['Gender']
    )

    gender_encoded_df = pd.DataFrame(
        gender_encoded,
        columns=encoded_columns,
        index=df_clean.index
    )

    print("\nEncoded Gender:")
    display(gender_encoded_df.head())


    df_clean = pd.concat(
        [
            df_clean.drop(columns=['Gender']),
            gender_encoded_df
        ],
        axis=1
    )

    numeric_features = [
        'AGE',
        'Urea',
        'Cr',
        'HbA1c',
        'Chol',
        'TG',
        'HDL',
        'LDL',
        'VLDL',
        'BMI'
    ]


    scaler = StandardScaler()

    df_clean[numeric_features] = scaler.fit_transform(
        df_clean[numeric_features]
    )

    scaling_check = pd.DataFrame({
        'Mean': df_clean[numeric_features].mean(),
        'Standard Deviation': df_clean[numeric_features].std()
    })

    print("\nScaling Verification:")
    display(scaling_check)

    print("\nFinal transformed dataset:")
    display(df_clean.head())

    print("\nFinal shape:", df_clean.shape)

    print("\nFinal columns:")
    print(df_clean.columns.tolist())
```
